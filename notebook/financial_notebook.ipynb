```python
# cleaning the messy data folder values (CSV)

messy["Rev"] = messy["Rev"].str.replace("$", "", regex= False)
messy["Rev"] = messy["Rev"].str.replace(",", "", regex= False)
messy["Rev"] = messy["Rev"].str.strip()
messy["Rev"] = pd.to_numeric(messy["Rev"])
```

```python
def clean_period(months):
    parts = months.split()
    if parts[0] == "Q1":
        return parts[1] + "-01-01"
    elif parts[0] == "Q2":
        return parts[1] + "-04-01"
    elif parts[0] == "Q3":
        return parts[1] + "-07-01"
    elif parts[0] == "Q4":
        return parts[1] + "-10-01"
    else:
        return months 

messy["Period"] = messy["Period"].apply(clean_period)
```

Chunking -> Embedding -> Store & compare 

```python
with open("../data/business_intelligence_market_trends_q3.txt") as f:
    content_1 = f.read()
print(content_1)
```

```python
chunks = content_1.split("\n\n")
print(chunks)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
vector = model.encode(chunks)
print(len(vector))
```

```python
import chromadb

client = chromadb.Client()
collection = client.create_collection(name = "Nexora_documents")
```

```python
collection.add(
    documents = chunks, 
    embeddings= vector.tolist(),
    ids = [f"chunks_{i}" for i in range (len(chunks)) ]
)
```

```python
results = collection.query(
    query_embeddings=model.encode(["What is the customer churn rate?"]).tolist(),
    n_results = 2
)
print(results)
```
